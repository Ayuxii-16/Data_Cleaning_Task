## Objective
To clean and prepare customer data by handling missing values, duplicates, incorrect data types, and inconsistent data.

```python
#Import Pandas
import pandas as pd
```

```python
#Load the dataset
df = pd.read_csv("messy_customer_data.csv")
```

```python
# Display the first 5 rows
df.head()
```

```python
# Check the dataset structure
df.info()
```

```python
# Check the number of rows and columns
df.shape
```

```python
df.isnull().sum()
```

```python
#Check for duplicated rows
df.duplicated().sum()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check duplicates again
df.duplicated().sum()
```

```python
# Fill missing values in numerical columns with the median
df = df.fillna(df.median(numeric_only=True))
```

```python
# Check if missing values remain
df.isnull().sum()
```

```python
# Check the data types of each column
df.dtypes
```

```python
# Convert date columns to datetime format
df["Signup_Date"] = pd.to_datetime(df["Signup_Date"], errors="coerce")
df["Last_purchase_date"] = pd.to_datetime(df["Last_purchase_date"], errors="coerce")
```

```python
# Check the date column data types
df[["Signup_Date", "Last_purchase_date"]].dtypes
```

```python
# Check for invalid or missing dates
df[["Signup_Date", "Last_purchase_date"]].isnull().sum()
```

```python
# Remove extra spaces from text columns
text_columns = ["Name", "Gender", "City", "email", "Phone_number", "Country"]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
# Display the cleaned dataset
df.head()
```

```python
# Check the final shape of the cleaned dataset
df.shape
```

```python
# Remove personal contact information before publishing
df_cleaned = df.drop(columns=["email", "Phone_number"])

df_cleaned.head()
```

```python
# Save the cleaned dataset
df_cleaned.to_csv("cleaned_customer_data.csv", index=False)
```

# Conclusion

The customer dataset was cleaned by:

- Removing duplicate records
- Handling missing values
- Converting date columns to the correct format
- Cleaning text values and removing extra spaces
- Removing personal contact information from the publishable dataset

The cleaned dataset is now ready for further analysis.
